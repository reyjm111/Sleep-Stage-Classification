```python
import mne
import numpy as np
import pandas as pd
```

```python
raw = mne.io.read_raw_eeglab("/Users/reymendoza/Desktop/sleep_files/sub-001/ses-001/sub-001_ses-001_task-sleep_acq-PSG_eeg.set", preload=True, verbose=False) # reading eeg file
raw.pick_channels(['ELA', 'ELB', 'ELC','ELT','ELE','ELI','ERA','ERB','ERC','ERT','ERE','ERI'], verbose=False) # selecting ear-eeg channels only

# Preprocess eeg
raw.filter(0.1, 40)
raw.notch_filter(60)
raw.resample(200)

df = pd.read_csv('/Users/reymendoza/Desktop/sleep_files/sub-001/ses-001/sub-001_ses-001_task-sleep_acq-scoring1_events.tsv', sep='\t') # reading annotation file
df['onset'] = df['onset'].round()

valid_codes = {1, 2, 3, 4, 5}
event_id = {
    'Wake': 1, 
    'REM': 2, 
    'N1': 3, 
    'N2': 4, 
    'N3': 5
}
events = []
epoch_len = 30.0
fs = raw.info['sfreq']

# collect onset times of epochs within the valid event codes
for i, code in enumerate(df['Scoring1']): 

    if code in valid_codes:

        onset_sample = int(i * epoch_len * fs)
        events.append([onset_sample, 0, code])

events = np.array(events, dtype=int)

epochs = mne.Epochs(raw, events, event_id=event_id, tmin=0, tmax=30 - 1/fs, baseline=None, preload=True) # create epochs object of shape (N, 0, 5)
```
